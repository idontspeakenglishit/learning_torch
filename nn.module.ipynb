```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
torch.manual_seed(1)
```

```python
x_train = torch.FloatTensor([[1], [2], [3]])
y_train = torch.FloatTensor([[2], [4], [6]])
```

```python
# 단순 선형 회귀 모델
model = nn.Linear(1,1) # input_dim=1,output_dim=1
print(list(model.parameters())) # W, b (참고, 모델 선언시 파라미터의 값은 랜덤으로 초기화 됨)
```

```python
optimizer = torch.optim.SGD(model.parameters(), lr=0.01) 
```

```python
nb_epochs = 2000
for epoch in range(nb_epochs+1):

    # H(x)
    prediction = model(x_train) # model(x_train) == model.forward(x_train)
    
    cost = F.mse_loss(prediction, y_train) # 파이토치에서 제공하는 평균 제곱 오차 함수

    optimizer.zero_grad()
    cost.backward()
    optimizer.step()

    if epoch % 100 == 0:
      print('Epoch {:4d}/{} Cost: {:.6f}'.format(
          epoch, nb_epochs, cost.item()
      ))
```

```python
new_var =  torch.FloatTensor([[4.0]])
pred_y = model(new_var) # forward 연산
print("훈련 후 입력이 4일 때의 예측값 :", pred_y) 
```

```python
x_train = torch.FloatTensor([[73, 80, 75],
                             [93, 88, 93],
                             [89, 91, 90],
                             [96, 98, 100],
                             [73, 66, 70]])
y_train = torch.FloatTensor([[152], [185], [180], [196], [142]])
```

```python
# 다중 선형 회귀 모델
model = nn.Linear(3,1) # input_dim=3, output_dim=1
print(list(model.parameters())) # [W_1, W_2, W_3], b
```

```python
# 단순과 다중 회귀 lr(학습률)의 수치 차이
# 입력 데이터 크기(x)로 인해 발생, 오차 함수 미분시 기울기 크기의 영향을 미침
# 즉, 오버슈팅으로 인한 기울기 발산을 방지하기 위함
optimizer = torch.optim.SGD(model.parameters(), lr=1e-5) 
```

```python
nb_epochs = 2000
for epoch in range(nb_epochs+1):
    prediction = model(x_train)

    cost = F.mse_loss(prediction, y_train)

    optimizer.zero_grad()
    cost.backward()
    optimizer.step()

    if epoch % 100 == 0:
      print('Epoch {:4d}/{} Cost: {:.6f}'.format(
          epoch, nb_epochs, cost.item()
      ))
```

```python
new_var =  torch.FloatTensor([[73, 80, 75]]) 
pred_y = model(new_var) 
print("훈련 후 입력이 73, 80, 75일 때의 예측값 :", pred_y) 
```

```python
# 단순 선형 회귀 모델 class
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(1, 1)

    def forward(self, x):
        return self.linear(x)
```

```python
model = LinearRegressionModel() # nn.Module.__call__() 실행시 forward를 리턴함으로써 상속을 통해 오버라이딩한 forward가 실행됨
```
